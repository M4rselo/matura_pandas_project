### Imporing Python libraries and creating dataframes for all given files

```python
import pandas as pd
import numpy as np
from charset_normalizer import from_path
import glob
```

```python
def kodowania_plikow(file_names):
    return {f_name: from_path(f_name).best().encoding for f_name in file_names}

f_nametab = glob.glob("*.txt")
print(kodowania_plikow(f_nametab))
```

```python
def setting_dataframes(file_names):
    return_dct = {}
    for f_name in file_names:
        return_dct[f_name] = pd.read_csv(f_name, sep="\t", encoding="cp1250")
    return return_dct

df_dct =  setting_dataframes(f_nametab)

print(', '.join([name.strip('.txt') for name in df_dct.keys()]))
```

```python
kraje, punkty, skoczek, skocznie, zawody = df_dct.values()
```

### Task 1

```python
def task_print(num):
    print(f"-------------TASK-{str(num)}-ANSWER-------------\n")
```

```python
def merge_inner(df1, df2, df1_col, df2_col):
    return df1.merge(df2, how='inner', left_on=df1_col, right_on=df2_col)
```

```python
task1_df = merge_inner(zawody, skoczek, 'kod_zawodnika', 'kod_skoczka')
task1_df = merge_inner(task1_df, kraje, 'kod_kraju', 'kod_kraju')

task1_df.head()
```

```python
task1_df.sort_values('punkty', ascending=True).head(3)
```

```python
ans_1 = task1_df.sort_values('punkty', ascending=True).iloc[0]

task_print(1)

print(ans_1['nazwisko_imie'], "|", ans_1['nazwa'], "|", ans_1['punkty'])
```

### Task 2

```python
task2_df = merge_inner(skoczek, kraje, 'kod_kraju', 'kod_kraju')
task2_df = task2_df.groupby(['nazwa']).count().rename(columns={'kod_skoczka': 'suma'})[['suma']]
min_2 = task2_df.suma.min()
max_2 = task2_df.suma.max()

ans_2 = task2_df[task2_df.suma.isin([min_2, max_2])]
task_print(2)
ans_2
```

### Task 3

```python
task3_df = merge_inner(zawody, skoczek, 'kod_zawodnika', 'kod_skoczka')
task3_df = merge_inner(task3_df, kraje, 'kod_kraju', 'kod_kraju')
task3_df['suma'] = task3_df['odl.1'] + task3_df['odl.2']

task3_df = task3_df.groupby(['nazwisko_imie', 'nazwa']).suma.sum().reset_index().sort_values('suma', ascending=False)

task3_df.head()
```

```python
ans_3 = task3_df.iloc[0]
task_print(3)
print(ans_3['nazwisko_imie'], "|", ans_3['nazwa'], "|", ans_3['suma'])
```

### Task 4

```python
task4_df = merge_inner(zawody, punkty, 'miejsce', 'miejsce').rename(columns={'punkty_y': 'punktacja'})
task4_df = merge_inner(task4_df, skoczek, 'kod_zawodnika', 'kod_skoczka')

ans_4 = task4_df.groupby(['kod_zawodnika', 'nazwisko_imie']).punktacja.sum().sort_values(ascending=False).reset_index()

task_print(4)

ans_4.head(3)
```

### Task 5

```python
task5_df = merge_inner(zawody, skocznie, 'skocznia', 'kod_skoczni')
task5_df = task5_df[task5_df['punkt_K'] > 170]
task5_df = merge_inner(task5_df, skoczek, 'kod_zawodnika', 'kod_skoczka')
task5_df = merge_inner(task5_df, kraje, 'kod_kraju_y', 'kod_kraju')
task5_df = merge_inner(task5_df, punkty, 'miejsce', 'miejsce').rename(columns={'punkty_y': 'punktacja'})

ans_5 = task5_df.groupby(['kod_zawodnika', 'nazwisko_imie', 'nazwa']).punktacja.sum().sort_values(ascending=False).reset_index()
ans_5.head()
```

```python
ans_5 = ans_5.iloc[0]
task_print(5)
print(ans_5['nazwisko_imie'], "|", ans_5['nazwa'], "|", ans_5['punktacja'])
```

### Task 6

```python
task6_df = merge_inner(zawody, skoczek, 'kod_zawodnika', 'kod_skoczka')
task6_df = merge_inner(task6_df, kraje, 'kod_kraju', 'kod_kraju')
task6_df = merge_inner(task6_df, skocznie, 'skocznia', 'kod_skoczni')

ans_6 = pd.pivot_table(task6_df, values='kod_zawodnika', columns='miasto', index='nazwa', aggfunc='count').replace(np.nan, 0)
task_print(6)
ans_6
```
